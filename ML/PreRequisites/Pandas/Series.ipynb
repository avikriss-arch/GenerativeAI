## Pandas Series is a column of data in a table like in Excel or Google sheets with label called as indexes attached to each value.

## Creating a Series in multiple ways -
## 1. List
## 2. Numpy Array
## 3. Dictionary
## 4. Custom Index
## 5 Scalar value

```python
# Series from List
country = ['India','China','Russia','Japan']
pd.Series(country)
```

```python
run = [47,98,32,100,156,12]
# pd.Series(run)
pd.Series(run,dtype='int32')
```

```python
subject = ['Maths','Science','English','Computer','Hindi']
marks = [91,89,81,100,92]
# pd.Series(subject,marks)
pd.Series(data=marks,index=subject,dtype='int32')
```

```python
# Series from Numpy Array
value = np.arange(10,21)
pd.Series(value)
```

```python
# Series with Dictionary
# Dictionary is another Data Structure that stores the data in Key:Value pair
# Just because it stores the data in a Key:Value pair so we don't have to explicitly mention the index

data = {'Maths':100,'English':82,'Hindi':85,'Science':93,'Social Science':68}
pd.Series(data)
```

```python
# Series from Custom Index
value = [10,20,30,40,50]
pd.Series(value, index=['a','b','c','d','e'])
```

```python
# Series from Scalar Value
pd.Series(10,index = ['Kitkat','Dairy Milk','Fuse','Eclairs'])
```
