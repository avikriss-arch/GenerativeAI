# Assignment 2: Loop Comntrol with Conditions (break & continue)
## Task 4

```python
daily = [200, 150, 0, 400, 50, -1, 300]

total_sales = 0
for value in daily:
    if value == -1:
        print(f"Corrupted data {value}, processing stopped")
        break

    if value == 0:
        print("A day with no sales, skipping")
        continue

    total_sales += value
    print(f"Added {value} and Running total: {total_sales}")

print(f"\nTotal sales: {total_sales}")

```
