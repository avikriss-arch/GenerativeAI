```python
# Understanding the above code (cell no. 529)
''' 
1. Understanding np.arange(mint.shape[0])
np.arange(mint.shape[0])
mint.shape[0] = the NUMBER OF ROWS in mint (e.g., if mint has 8 months of data, this is 8)
np.arange(8) generates: [0, 1, 2, 3, 4, 5, 6, 7]

>this creates a simple numbered position for each month on the x-axis — position 0 for the first month,
position 1 for the second month, and so on.

2. Why Subtract/Add 0.2 to Each Series
This is the key trick that makes bars sit SIDE-BY-SIDE instead of directly on top of each other:
plt.bar(np.arange(mint.shape[0]) - 0.2, mint.values, ...)     # shifted LEFT
plt.bar(np.arange(dark.shape[0]), dark.values, ...)             # stays CENTER
plt.bar(np.arange(peanut.shape[0]) + 0.2, peanut.values, ...)   # shifted RIGHT


Visual Diagram (imagine Month 1, Month 2 on x-axis)
Position:    -0.2  0   0.2      0.8  1   1.2
              |    |    |        |   |    |
             mint dark peanut   mint dark peanut
            (Month 1 group)    (Month 2 group)

'''
```

# Stacked Bar Chart
## --> Instead of keeping the bars side by side direction, we can keep the bars in the stacked manner i.e. One on Top of other, 
## and in order to figure out which category is having maximum sale, we can have a look over the length of the Bar

# Histogram
## -->A histogram shows the distribution of ONE numeric variable. It splits the range of values into equal-width intervals called bins, counts how many data points fall into each bin, and draws a bar for each count.
## -->This is univariate analysis

```python
import matplotlib.pyplot as plt

ages = [18, 19, 21, 22, 22, 23, 24, 24, 24, 25, 26, 27, 28, 30, 31, 35]

plt.hist(ages)
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
```

```python
# Key Parameters
# 1.bins (the most important one)
plt.hist(ages, bins=5)      #5 equal-width bins
```

```python
plt.hist(ages, bins=[15,20,25,30,35])       # custom bins ages
```

```python
# 2. color and edgecolor
plt.hist(ages, bins=5, color='skyblue', edgecolor='red')     # edgecolor draws a border around each bar, so adjacent bins are easier to tell apart.
```

```python
plt.hist(ages, orientation="horizontal")
```

```python
group_a = [20, 22, 23, 25, 25, 27, 30]
group_b = [24, 26, 28, 29, 30, 32, 35]

plt.hist(group_a, bins=5, alpha=0.6, color="blue", label="Group A")         #alpha ranges from 0 (invisible) to 1 (solid). It matters when overlaying multiple histograms.
plt.hist(group_b, bins=5, alpha=0.6, color="red", label="Group B")
plt.legend()            #alpha lets both histograms show through each other, and label plus plt.legend() identifies them.
plt.show()
```

# Pie Chart
## --> It is used for Bivariate Analysis and it's being used for relating Categorical and Numerical column each other but in Pie Chart we will look to percentage distribution (Univariate Analysis) and Bivariate Analysis (Categorical-Numerical) as in terms of Category by Category Percentage Distribution in Aggregation of Numerical Data.

```python
dh
```

```python
plt.figure(figsize=(10,7))
plt.pie(dh, labels=dh.index, autopct='%0.1f%%', colors=['red','green','blue'])         # autopct used for percentage, labels used for lableing as per indexes like Unhealthy, Moderate, Healthy,Others 
plt.legend()                                    
plt.show()
```

```python
plt.figure(figsize=(10,7))
plt.pie(dh, labels=dh.index, autopct='%0.1f%%', explode=[0.1,0,0,0], shadow=True)         # autopct used for percentage, labels used for lableing as per indexes like Unhealthy, Moderate, Healthy,Others 
plt.legend()                                    
plt.show()
```
