# OS Library 

## OS Library is able to interact with our Operating System and can let us work with multiple Directories and Files.

## getcwd() --> It is capable of providing us with an idea about the current working directories.
## getcwd --> Stands for "get current working directory"

```python
import os
```

```python
# I want to know what is my current working directory

current_working_directory = os.getcwd()
print("Current working directory is: ",current_working_directory)
```

## mkdir() --> make directory --> It will be helping us out with creating the directory

```python
new_path = 'New_folder'
os.mkdir(new_path)
print(f'New directory {new_path} created successfully!')
```

```python
print(os.mkdir('New_test_dir'))      # Directory could be created with this same line, same logic
```

## listdir() --> List Directories --> Listing all of the Directories present

```python
All_Directories = os.listdir('.')
print(All_Directories)
```

## os.path.join() --> It will be joining multiple paths together

```python
Dir_name = 'Folder_Data'
File_name = 'File_Data'
Full_Path = os.path.join(Dir_name,File_name)
print(f'Full path is {Full_Path}')
```

```python
# I want to join multiple path with the current path

Complete_path = os.path.join(os.getcwd(),Dir_name,File_name)
print(f'Complete path is: ',Complete_path)
```

## os.path.exists --> To check given path exists or not 

```python
given_path = '/Users/avinashkumar/GenAI_Course/Python/app.py'        # copy the path and paste here

if os.path.exists(given_path):
    print('Given path exists')
else:
    print('Given path does not exists')
```

## os.path.isfile() --> This checks that whether if the current path is a File or not.
## os.path.isdir() --> This checks that whether if the current path is a Directory or not.

```python
path = 'total_data.txt'

if os.path.isfile(path):
    print('The current path belongs to a File')
elif os.path.isdir(path):
    print('The current path belongs to a Directory')
else:
    print('The current path neighther belongs to a File nor to a Directory')
```

```python
paths = 'New_folder'     # /Users/avinashkumar/GenAI_Course/Python/Modules and Packages/File Handling/New_folder

if os.path.isfile(paths):
    print('The current path belongs to a File')
elif os.path.isdir(paths):
    print('The current path belongs to a Directory')
else:
    print('The current path neighther belongs to a File or a Directory')
```

## os.path.abspath() --> It provides us with the complete path or the Absolute Path of the File or Directory

```python
# Find a complete path of the given file 'total_data.txt'
# total_data.txt <--> It's absolute path
# /Users/avinashkumar/GenAI_Course/Python/Modules and Packages/File Handling/total_data.txt   <--> It's absolute path

file_name = 'total_data.txt'

absolute_path = os.path.abspath(file_name)

print(f'The complete path of the file "total_data.txt" is {absolute_path}')
```
