# input()

```python
name = input("Enter your name: ")  # input() function is used to take input from the user. It always returns a string.
```

```python
print(name)
```

## input function only takes the input in a string format until and unless we are going to explicitly take the input in interger or other format.

```python
Age = input("Enter your age:")
```

```python
print(type(Age))  # type() function is used to check the data type of a variable. It will return <class 'str'> because input() function always returns a string.
```

```python
Age = int(input("Enter your age: "))  # input() function is used to take input from the user. It always returns a string. We can typecast it to int.
print(type(Age))  # type() function is used to check the data type of a variable. It will return <class 'int'> because we have typecasted it to int.
```

```python
Age = float(input("Enter your age: ")) 
print(type(Age))
```
