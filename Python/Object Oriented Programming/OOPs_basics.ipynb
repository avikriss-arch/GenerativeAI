## Types of Constructor

## 1. Default Constrcutor
## 2. Parameterized Constructor

## 1. Default COnstructor --> It does not have the Parameters inside it and it is only having the self keyword inside it. It assigns a default value to all of the Object.

```python
class Dog:
    def __init__(self):
        self.name='Labrador'
        self.color='Brown'

    def show_info(self):
        print(f'This Dog is {self.name} and the color of this dog is {self.color}')
```

```python
dog1 = Dog()     # object one created 
```

```python
dog1.show_info()
```

```python
dog2 = Dog()      # Object two created
```

```python
dog2.show_info()
```

## Parameterized Constructot --> It consists of Parameters inside it so that every Object can have a specific attributes associated to it. It allows us pass arguments while creating an object.

```python
class Dogs:
    def __init__(self,breed,color):     # parameterise constructor
        self.breed=breed
        self.color=color

    def show_info(self):                 # instance method
        print(f'The dog is {self.breed} and color is {self.color}')
```

```python
dog_object1 = Dogs('Lebrador','Brown')
print(dog_object1.show_info())
```

```python
dog_object2 = Dogs('German Shepherd','Brown and Black')
print(dog_object2.show_info())
```

## Types of Method

## 1. Instance Method 
### --> Instance method operates on Object Attribute. 
### --> Instance method is going to use self keyword to access and modify object attributes.
### --> These methods are called using an Object. 
## 2. Class Method
### --> It is the method of the whole class or the whole blueprints which will remain the same for all of the Objects then such kind of method is referred by a Class Method.
### --> In Class Method, these methods work with the class level attributes.
### --> They are going to use 'cls' keyword. 
### --> They are called using the Class name as well.
### --> For specifying that the current method is a class method we have to use the abbreviation as @classmethod
## 3. Static Method
### --> Static Methods do not use 'self' or 'cls' .
### --> They work like normal function but belong to a Class.
### --> They are going to be designated by a abbreviation @staticmethod.
### --> Static method can also be called by the class name 


```python
# Instance method
class Avengers:
    def __init__(self, name , power):
        self.name=name
        self.power=power
    
    def intro(self):
        print(f'{self.name} has {self.power} power')

avenger_object = Avengers('Iron Man','Intelligence')
avenger_object2 = Avengers('Thor' ,'God of Thunder')

print(avenger_object.intro())
print(avenger_object2.intro())
```

```python
# Class method
class Marvel:

    origin = 'Shield'

    def __init__(self,name, power):
        self.name=name
        self.power=power

    @classmethod
    def class_method(cls):
        print(f'My origin is {cls.origin}')

    def instance_method(self):
        print(f'Hi, I am {self.name} and my power is {self.power}')

Iron_man_obj = Marvel('Thor','God of thunder')
print(Iron_man_obj.instance_method())
print(Iron_man_obj.class_method())
    
```

```python
print(Iron_man_obj.name)
print(Iron_man_obj.power)
```

```python
# Static Method
class Marvel:

    @staticmethod                                           #static method
    def static_method():
        print('All Avengers must protect the Earth')

    def __init__(self,name,power):                          #constructor and we can any name instead of 'self' like def __init__(avi,name,power)
        self.name=name
        self.power=power

    origin='Stanley'

    @classmethod                                            #class method
    def class_method(cls):
        print(f'The Avengers characters are created by {cls.origin}')

    def instance_method(self):                                      #instance method
        print(f'I am {self.name} and my power is {self.power}')



thor_obj = Marvel('Thor','Thunder')
```

```python
print(thor_obj.static_method())
```

```python
print(thor_obj.class_method())
```

```python
print(thor_obj.instance_method())
```

```python
print(Marvel.static_method())      # calling static method using class name
```
