# Encapsulation

## Encapsulation is a way of wrapping data (variables and methods) inside a capsule or a class as a single unit.

## It restricts the direct access to some of the objects components which is a means of preventing accidental inference and misuse of the data.

## Key concepts of Encapsulation
## 1. Public members
## 2. Protected members
## 3. Private members 
## getter and setter method

## 1. Public Member
### --> A Public member is accessible from anywhere or attributes are accessible outside of the class as well and these members are accessible from derived class as well.
### --> Public member attribute is directly declared 

```python
class Ironman:
    def __init__(self,name,weapon):
        self.name=name
        self.weapon=weapon

    def intro(self):
        print(f"Hello my name is {self.name} and my weapon is {self.weapon}")
```

## dir() -->It takes the input as the Object name and gives us this idea that which member is public or protected or private

```python
ironman_obj = Ironman('Iron Man','Mark 82 Suit')
```

```python
print(ironman_obj.intro())
```

```python
dir(ironman_obj)
```

## Private Members 
### -->A Private member is not accessible outside the class and not even in the derived class and in order to have the access over these private members or to bring some modification on these members we use 'getter' and 'setter' method.
### -->Private members are declared by using double underscore (__) before the name.

```python
class Marvel:
    
    def __init__(self,name,phone):
        self.name=name
        self.__phone=phone         # private attribute using double underscore __
```

```python
mar_obj = Marvel("Thor","9742672348")
```

```python
mar_obj.name
```

```python
mar_obj.phone           # it will throw AttributeError becuase phone is private.
```

```python
dir(mar_obj)
```

## getter method
### -->it will be helping us out with accessing the private data 

```python
class Captain:
    def __init__(self,name,address):
        self.name=name
        self.__address=address

    #getter method
    def accessing_add(self):
        print(f"My address is {self.__address}")

captain = Captain("Captain America","New York")
```

```python
captain.name
```

```python
captain.__address       # it will throw AttributeError becuase it is a private attribute so we have handle it by using Getter method
```

```python
captain.accessing_add()         #using getter method direct called by object instead of attribute by object
```

```python
dir(captain)
```

## Setter Method
### --> This method is used for modifying our private data

```python
class Animal:
    def __init__(self,name,sound):
        self.name=name
        self.__sound=sound
    
    def get_sound(self):            #getter method
        print(f"The Animal is going to sound as: {self.__sound}")

    def set_sound(self,new_sound):      #setter method
        self.__sound=new_sound
        print(f"New sound for the Animal is: {self.__sound}")    
```

```python
animal_obj = Animal("Bruno","Woof Woof")        # object creation
```

```python
animal_obj.name
```

```python
animal_obj.sound            #this will throw AttributeError
```

```python
animal_obj.__sound          #this will also throw AttributeError
```

```python
# getter method to have access over this data
animal_obj.get_sound()
```

```python
# setter method helps to modify the data
animal_obj.set_sound("Meow Meow")
```

## Protected Member
### -->Protected member is a data which is not accessible outside the class but the protected data is going to be accessible by derived/child class.
### --> For having the protected member you can mention your protected member as single underscore(_) or by _attribute 

```python
class Grandparent:
    def __init__(self,name,lockerkey):
        self.name=name
        self._lockerkey=lockerkey             # used single underscore as prefix to make the lockerkey Protected member

    def get_locker(self):                   # getter method
        print(f"The locker key info is: {self._lockerkey}")

class Parent(Grandparent):
    def __init__(self,name,lockerkey):
        super().__init__(name,lockerkey)
        
    def child_reveal(self):
        print(f"The locker key information is: {self._lockerkey}")    


grandpa = Grandparent("Avinash","574387")
```

```python
grandpa.name
```

```python
grandpa.lockerkey
```

```python
grandpa._lockerkey
```

```python
dir(grandpa)
```

```python
grandpa.get_locker()
```

```python
grandpa.child_reveal()
```

```python
parent = Parent("Avi","4326")
```

```python
parent.name
```

```python
parent._lockerkey
```

```python
parent.lockerkey
```

```python
parent.get_locker()         # grandparent class
```

```python
parent.child_reveal()          #parent class
```

```python
# Bank Account System
class BankAccount:
    def __init__(self, account_holder, balance):
        self.account_holder = account_holder
        self.__balance = balance                  # private attribute (encapsulation)
    
    def get_balance(self):                          #getter method
        #print(f"Total balance is: {self.__balance}")
        return self.__balance                       # Returns the value, does not print it

    def set_deposit(self, amount):                  #setter method
        if amount > 0:
            self.__balance += amount
            return f"{amount} amount diposited \ncurrent balance is: {self.get_balance()}"       
        else:
            return "Deposited amount must be positive"

    def set_withdrawl(self, amount):                #setter method
        if 0 < amount <= self.__balance:
            self.__balance -= amount
            return f"Withdrawl amount is: {amount} \nRemaining balance is {self.get_balance()}"
        else:
            return "Insufficient amount"    
```

```python
acc = BankAccount("Avinash",60000)
```

```python
acc.get_balance()
```

```python
print(acc.set_deposit(5000))
```

```python
print(acc.set_withdrawl(3000))
```
