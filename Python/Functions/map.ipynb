# Map()

## Map Function applies a given function operation to all items in an Iterable or a List and returns a Map object (an Iterator)

## Map Function is particularly useful for transforming the data or applying an operation on all items of the Iterable 

## Map() is going to help us out with avoiding unnecessary lines of code and we can apply an operation on all values or all items in an Iterable and this is a sign of a good programmer.

## Syntax for Map() Function is:

### list(map(function , iterable))

```python
# Squaring all itens of a list using normal procedure
# list --> [1,4,5,3,6,7,8,9]
# [] --> [1,16,25....81]

def square(list_items):
    empty_list = []
    for item in list_items:
        square_items = item**2
        empty_list.append(square_items)
    print(empty_list)
    
```

```python
square([1,4,5,3,6,7,8,9])
```

```python
# Similar problem using Map()

print(list(map(lambda x:x**2 , [1,4,5,3,6,7,8,9]))) 
```

```python
# Map function with the basic function

def sqaure_num(num):
    return num**2
```

```python
print(list(map(sqaure_num ,[1,4,5,3,6,7,8,9])))

# Whenever you are going to pass the general function inside our map() function as the parameter then you don't have to use parenthesis ()
```

## While mapping the value of one Iterable with another Iterable we can actually apply the operation in between the values of the two iterables in synchronous way
### [1,2,3,4] <---> [5,4,7,9] ---> Addition ---> [6,6,10,13]



```python
# Map Function with multiple iterable

num1 = [1,2,3,4]
num2 = [5,4,7,9]

print(list(map(lambda x,y:x+y , num1 , num2)))
```

```python
# Map() to convert all the String value items in an iterable into an integer

map_value = list(map(int , ['1','3','5','7']))
```

```python
print(map_value)
```

```python
# convert the employee name in uppercase 

emp_name = ['avinash' , 'ashutosh' , 'ravi' ,'ujjwal' , 'pratyush']

employees = list(map(str.upper , emp_name))
print(employees)
```

```python
# Retrieve the Employee age from the dictionary
# step 1 --> create a dictionary
# step 2 --> create a function and return required details(age)
# step 3 --> create map()
Employee_details = [{'Name':'Avinash' , 'Age':24 , 'Qualification':'B.Tech'},
                    {'Name':'Shelly' , 'Age':25 , 'Qualification':'B.Com'},
                    {'Naem':'John' , 'Age':29 , 'Qualification':'BBA'}
                    ]
```

```python
def age_retrieval(list_items):
    return list_items['Age']
```

```python
print(list(map(age_retrieval , Employee_details)))
```
