## Shallow Copy --> Shallow Copy is being used for creating a copy of the Original Dictionary in a way such that the changes that we are going to make with the Original Dictionary won't be getting reflected into our copied Dictionary and Shallow Copy is being done by copy() method

```python
dictionary1={"Name":"Avi", "a":23, "b":56}
```

```python
dictionary2=dictionary1
```

```python
dictionary2
```

```python
# Changing the value in dictionary1
dictionary1["a"]=45
```

```python
dictionary1
```

```python
dictionary2     # changes have been also reflected in dicitonary2
```

```python
dictionary3=dictionary1.copy()
```

```python
dictionary3
```

```python
dictionary1['a']=6
```

```python
dictionary1       # change has been reflected
```

```python
dictionary2      # change has been reflected
```

```python
dictionary3     # change has not been reflected
```

## Nested Dictionary

### Nested Dictionary will be used for keeping a Dictionary inside another Dictionary

```python
# I want store the information for multiple students like their name, marks, grada
# syntax dict={{"key1":"value1" , "key2":"value2"},{"key1":"value1" , "key2":"value2"}}
nested_dictionary={"student1":{"name":"Avi" , "marks":65 , "grade":"B"} , 
 "student2":{"name":"Ravi" , "marks":89 , "grade":"A"},
 "student3":{"name":"Divya" , "marks":99 , "grade":"A+"}}
```

```python
nested_dictionary
```

```python
# Iterate over these Nested Dictionary
for outer_key,outer_value in nested_dictionary.items():
    print(f"Outer key is: {outer_key} and Outer value is: {outer_value}")
    for inner_key,inner_value in outer_value.items():
        print(f"Inner key is: {inner_key} and Inner value is: {inner_value}")
```

## Dictionary Comprehension

### Dcitionary Comprehension helps us with using Looping Statement along with Conditional statement inside our Dictionary

## Syntax for Dictionary Comorehension

## {expression for item in iterable if condition}

```python
comprehensive={x**2 for x in range(11)}
```

```python
comprehensive
```

```python
comprehensive={x:x**2 for x in range(11)}
```

```python
comprehensive
```

```python
even_dic={x:x**2 for x in range(11) if x%2==0}
```

```python
even_dic
```
