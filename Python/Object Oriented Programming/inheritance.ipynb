# Pillars of Object Oriented Programming

## Inheritance
### -->Inheritance is a Fundamental concept in Object Oriented Programming because of which the child class is able to inherit the Properties or Attributes and Funcationalities of the parent class.
### -->Just like a child is going to inherit the properties from parent and parent is going to inherit the properties from the grandparent and definitely in this hierarchy the granchild can have features different from the parent and grandparent as well.

## Types of Inheritance in Python
### 1. Single Inheritance
### 2. Multi-level Inheritance
### 3. Multiple Inheritance

## 1. Single Inheritance
### -->The Features or the Attributes from one parent class will be getting inherited into one child class.
### Parent Class (Properties --> Attributes and Functionalities)
### |
### |
### Child Class

```python
class Vehicle:                                  # Parent class
    def __init__(self,brand,speed):
        self.brand=brand
        self.speed=speed

    def start_engine(self):
        print(f"{self.brand}'s engine has started")

class Car(Vehicle):                             # Child class--> inherits from Vehicle
    def __init__(self,brand,speed,doors):
        super().__init__(brand,speed)           # Calls Vehicle's __init__
        self.doors=doors

    def open_trunk(self):
        print(f"{self.brand}'s trunk is open")

my_car = Car("Toyota",'140kmph','4')
my_car.start_engine()
my_car.open_trunk()
    

```

## 2. Multi-Level Inheritance
### -->The properties or Methods and Attributes of the grandparent class can be inherited by the parent class and then properties of the parent class will be inherited by the child class. In a proper hierachy, the Inheritance process is going to work for use.

```python
# Multi-Level Inheritance example 1
class Employee:
    def __init__(self,name):
        self.name=name

    def work(self):
        print(f'{self.name} is working')

class Manager(Employee):
    def __init__(self,name,team_size):
        super().__init__(name)
        self.team_size=team_size

    def manage_team(self):
        print(f'{self.name} is managing a team of {self.team_size}')

class SeniorManager(Manager):
    def __init__(self,name,team_size,department):
        super().__init__(name,team_size)
        self.department=department

    def set_strategy(self):
        print(f'{self.name} is setting strategy for {self.department}')

sm_object = SeniorManager("Narendra","18","Engineering")
sm_object.work()                    # from Employee(grandparent)
sm_object.manage_team()             # from Manager(parent)
sm_object.set_strategy()            # defined in SeniorManager itself


        
```

```python
# Multi-Level Inheritance example 2
class Ticket:
    def __init__(self,ticket_id):
        self.ticket_id=ticket_id

    def log(self):
        print(f'Ticket #{self.ticket_id} logged')

class ProductionTicket(Ticket):
    def __init__(self,ticket_id,severity):
        super().__init__(ticket_id)
        self.severity=severity

    def escalate(self):
        print(f'Ticket #{self.ticket_id} escalated due to {self.severity} severity')

class PaymentTicket(ProductionTicket):
    def __init__(self,ticket_id,severity,gateway):
        super().__init__(ticket_id,severity)
        self.gateway=gateway

    def check_gateway_status(self):
        print(f'Checking {self.gateway} gateway status for ticket #{self.ticket_id}')


pt = PaymentTicket(501,"High","Juspay")

pt.log()
pt.escalate()
pt.check_gateway_status()
```

```python
# Multi-Level Inheritance example 3
class Agent:

    def __init__(self,name):
        self.name=name
    
    def think(self):
        print(f'{self.name} is processing the request')

class ToolUsingAgent(Agent):

    def __init__(self,name,tools):
        super().__init__(name)
        self.tools=tools

    def use_tool(self,tool_name):
        print(f'{self.name} is using tool {tool_name}')

class RAGAgent(ToolUsingAgent):

    def __init__(self,name,tools,vector_db):
        super().__init__(name,tools)
        self.vector_db=vector_db

    def retrieve_and_response(self,query):
        print(f'{self.name} searches {self.vector_db} for {query}')




agent = RAGAgent('SupportBot',['search','calculator'],'Chroma')       
agent.think()           # from Agent
agent.use_tool("search")     # from ToolUsingAgent
agent.retrieve_and_response("refund policy")
```
