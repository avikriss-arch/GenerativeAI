# Abstraction

## -->Abstraction is the concept of hiding the implementation details and only showing the necessary features of the object.
## -->This helps in reducing programming complexity and efforts
## -->Abstraction allows us to work with an object with what an object does rather than how it does it.

## How can we achieve abstraction in Python
## 1.Abstract Class
### -->An class that can not be instantiated or that means for an abstract class we can't create the object of this class 
## 2.Abstract Method
### -->Abstract method just because it can't accessed by the object of the abstract class itself that's why we have to call this abstract method in our child class.And, the implementation of the Abstract method is not provided inside the Abstract class but must be provided in the child class.
### -->Abstract method is being defined by @abstractmethod

```python
# 1. Example of Abstraction : Agent system

from abc import ABC, abstractmethod

class Agent(ABC):               # ABC =  Abstract Base 
    
    @abstractmethod
    def respond(self):
        pass                    # no implementation here - forces subclass to define it 

    @abstractmethod
    def get_name(self):
        pass



class SupportAgent(Agent):
    def respond(self,query):
        return f"Resolving your issue: {query}"

    def get_name(self):
        return "SupportAgent"



class SalesAgent(Agent):
    def respond(self,query):
        return f"Here's a recommendation for: {query}"

    def get_name(self):
        return "SalesAgent"


# This would FAIL - you can not create an object directly from an abstract class:
# agent = Agent()     # TypeError: Can't instantiate abstract class


support = SupportAgent()
print(support.respond("Payment issue"))             # works fine - implements the "contract"

sales = SalesAgent()
print(sales.get_name()) 
```

```python
# 2. Example : Real-World Everyday Abstraction — Car Class

class Car:
    def start_engine(self):
        self.__check_fuel()
        self.__ignite_spark_plug()
        self.__engage_starter_motor()
        print("Engine Started!")

    def __check_fuel(self):
        print("Checking fuel level...!")        # private method - hidden internal detail

    def __ignite_spark_plug(self):
        print("Igniting spark plugs....")       # private method - hiddent internal detail

    def __engage_starter_motor(self):
        print("Engaging starter motor...!")     # private method - hidden internal detail

    

my_car = Car()
my_car.start_engine()                           # user only call ONE simple method
# my_car.__check_fuel()                         # This will not work directly - hidden detail
    

```

```python
# 3. Example : LLM API Wrapper 

class LLMClient:
    def __init__(self,api_key):
        self.__api_key = api_key

    def ask(self, question):
        # Internally this method might handle: building the request payload, adding the authentication header, calling the actual API,
        # parsing the JSON response, handling errors, retrying on failures - All hidden
        return self.__call_api(question)

    def __call_api(self,question):
        # complex internal logic here
        # requests.post(url, headers=..., json=...), error handling, etc.
        return f"Answer to {question} defined as- \nRAG stands for Retrieval-Augmented Generation.\nIt is an artificial intelligence framework that improves the accuracy and reliability of large language models (LLMs) \nby fetching facts from an external, trusted database before generating a response."



client = LLMClient(api_key="sk-123")
print(client.ask("What is RAG"))               # simple interface - one method call 

# a = input("Ask your question")        # can take the input from the use as well
# print(client.ask(a))            


```

## Interview-ready one-liner:
### "Abstraction hides complex implementation details behind a simple interface, so users interact with WHAT an object does, without needing to know HOW it works internally — commonly implemented in Python using abstract base classes (ABC, @abstractmethod) or by hiding internal logic behind simple public methods."
