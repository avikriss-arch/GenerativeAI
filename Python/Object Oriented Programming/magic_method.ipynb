# Magic Methods

## Magic methods are also called as Dunder method because they have Double Underscore __ before and after their name and they are special method in Python that allow us to customise how objects behave

## These method can help us out:
## 1. Object Initialization (__init__)
## 2. String Representation (__str__ , __repr__)
## 3. Arithmetic Operations (__add_ , __sub__ ,etc)
## 4. Comparison Operations (__eq__ , __lt__)
## 5. Length, Indexing, Iteraton (__len__ , __getitem__ , etc) 

## Why to use Magic Method
### -->Makes objects behave like built-in type (eg - lists, integers)
### -->Improves readability by allowing basic operations object1 + object2
### -->Simplfies code by replacing verbose function call with natural syntax
