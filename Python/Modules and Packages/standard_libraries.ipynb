## CSV Library --> Most of the time we are going to work with csv file so that's why knowledge with csv library functionality matters

```python
import csv
```

```python
with open('example.txt' , mode='w' , newline='') as file:
    writer = csv.writer(file)
    writer.writerow(['Name' , 'Age'])
    writer.writerow(['Avinash' , 24])
```

```python
with open('example.txt' , mode='r') as file:
    reader = csv.reader(file)
    for row in reader:
        print(row)
```

## Time Library
### Time Library provides us the current time and if you want your Interpreter to sleep for seconds you can do it by using the Time Library

```python
import time
```

```python
print(time.time())
```

### If you are writing a piece of code and you want your Interpreter to sleep for a number of seconds then you can make the use of sleep function 

```python
print(time.time())    # this will run first then
time.sleep(5)         # Iterpreter will sleep for 5 seconds then 
print(time.time())    # this code will run after 5 seconds
```

## JSON Library 
## --> When we are going to work with Nosql Database then the Data inside the Nosql Database is going to be there in the JSON format and along with that when we are going to work with the datasets as well then again these dataset will be consisting of data in the JSON format itself

## When we get the JSON data , we get the JSON string format so by using JSON library we can convert our string data into dictionary format for serializing our data.

```python
import json
```

```python
dictionary = {'Name':'Avinash' , 'Age':25}
```

```python
json_str = json.dumps(dictionary)
```

```python
print(json_str) 
print(type(json_str))      # type converted to string 
```

```python
json_dictionary = json.loads(json_str)
```

```python
print(json_dictionary)
print(type(json_dictionary))         # type converted back to dictionary
```

## Regulary Expression 
## --> It is used to retrieve info based on a specific pattern or if the pattern matches

```python
import re
```

```python
text = 'Hello my name is Avinash and my password is Avinash@12345'
```

```python
pattern = r'\d+'
match = re.search(pattern,text)
print(match.group())
```
