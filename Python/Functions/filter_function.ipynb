# Filter Function 

## Filter function is going to retrieve only those items which are meeting a certain condition in an Iterable

### The Filter Function constructs an Iterator from elements of an Iterable for which a condition remains true or for which a Function returns True value

## Syntax for Filter Function

## list(filter(function, iterable))

```python
# To retrieve only those values which are Even
# step 1 --> create a list
# step 2 --> create a function
# step 3 --> use filter(function, iterable) function

iterable_list = [2,1,24,3,4,56,76,57,23,19,98,100]
```

```python
def even_function(values):
    if values%2==0:
        return True
```

```python
final_value = list(filter(even_function,iterable_list))
```

```python
print(final_value)
```

```python
# OR
print(list(filter(even_function,iterable_list)))
```

```python
# Similar problem using Lambda Function
print(list(filter(lambda x:x%2==0,iterable_list)))
```

```python
# I want to retrieve those value which are greater than 40 (use the same iterable_list)

print(list(filter(lambda x:x>40, iterable_list)))
```

```python
# Check with map function what value comes for the same logic 
print(list(map(lambda x:x%2==0,iterable_list)))
```

```python
# i want to retrieve those values which are greater than 40 and odd number

print(list(filter(lambda x:x>40 and x%2!=0, iterable_list)))
```

```python
# i want to retrieve those values which are greater than 40 and odd number

print(list(filter(lambda x:x>40 and x%2==0, iterable_list)))
```

```python
# I want to retrieve Employees whose age are geater than 30
# Step 1 --> create dictionary
# Step 2 --> create function
# Step 3 --> use filter(function,iterable)

Employees =[{'Name':'Avinash' , 'Age':25 , 'Gender':'Male'},
            {'Name':'Radha' , 'Age':31 , 'Gender':'Female'},
            {'Name':'Kishan' , 'Age':38 , 'Gender':'Male'},
            {'Name':'Ram' , 'Age':21 , 'Gender':'Male'},
            {'Name':'Shyam' , 'Age':45 , 'Gender':'Male'}]
```

```python
def age_filter(employee):
    if employee['Age']>=30:
        return True
```

```python
print(list(filter(age_filter,Employees)))
```
