```python
d = input("Enter denominatory")    # input coming as string value 

try:
    result = 1/d
    print(result)
except TypeError as ex3:        # this will come when we provide string value in this case as we have not converted 'input' in 'int'
    print(ex3)
```

## If we are getting an Error and we are not aware that which specific error type it is then we can use the primary exception class which is 'Exception'

## The Parent Exception type must be mentioned at the end or last as it is the Parents Class, so only if the previous exception class will catch the error then this final one will be able to catch it

## try --> try block is going to have the operation where we can except an error to occur.
## except --> It is used to handle the error in as way such that the program execution should not stop.
## else --> It will be consisting of the code which we want to run if the try block operation is working without throwing any Error.
## finally --> It gets executed once we are done with everything and even if we are getting an error or not getting an error, still the code inside the finally block gets executed.


## if Try block --> code is throwing an Error --> Except block --> code will be running 
## if Try block --> code is not throwing an Error --> Else block --> code will be running 

```python
den = int(input('Enter denominator'))

try:
    result = 1/den
except Exception as exc:
    print(exc)
else:
    print(f'Division for your denominator is: {result}')
```

```python
try:
    den1 = int(input("Enter first denominator"))       # taking the input as an Int
    result1 = 1/den1

    den2 = input('Enter second denominator')        # taking the input as a String, TypeError will come, The TypeError is not mentioned so the error will go in Exception
    result2 = 1/den2

except ZeroDivisionError as ex1:
    print('Ex1:',ex1)

except NameError as ex2:
    print('Ex2: ', ex2)

except Exception as exc:
    print("Exception: ", exc)

else:
    print(f'The division of first denominator is {result1}')
    print(f'The division of second denominator is {result2}')

finally:
    print('Program has executed successfully')    
```
