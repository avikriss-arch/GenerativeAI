# Conditional Statements

## Condtional statement works like real life scenario where if a certain condition is meeting then a statement of code or block of code gets executed.

```python
# Eligibility for driving on meeting a certain condition
# Conditional statement --> if statements where if a condition is meeting then a block of code gets executed

age = int(input("Enter your age: "))

if age>=18:
    print("You are eligible for Driving")

else:
    print("Sorry! You are not eligible for Driving because your age is", age)
```

## Conditional statements
## 1. if statement
## 2. elif statement
## 3. else statement

```python
# Figure out who is the tallest between the two friends
Avi = int(input("Enter Avi's height: "))
Shelly = int(input("ENter Shelly's height: "))

if Avi>Shelly:
    print("Avi is taller than Shelly")
else:
    print("Shelly is taller tha Avi")
```

```python
# Figure out who is tallest amonth three friends
friend1 = int(input("Please, Enter friend1 height: "))
friend2 = int(input("Please, Enter friend2 height: "))
friend3 = int(input("Please, Enter friend3 height: "))

if friend1>friend2>friend3:
    print("Friend1 is taller among all")
elif friend2>friend1>friend3:
    print("Friend2 is taller among all")
else:
    print("Friend3 is taller amoung all")
```

# Nested Conditional Statements

## Nested Conditional Statements involve if statements inside another if statement 

```python
# Whether if the current year is a leap year or not
year = int(input("Enter the year: "))

if year%4==0:
    if year%100==0:
        if year%400==0:
            print("Leap year")
        else:
            print("Not Leap year")
    else:
        print("Leap year")
else:
    print("Not a Leap year(4)")
```
