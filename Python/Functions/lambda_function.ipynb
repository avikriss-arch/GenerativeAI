# Lambda Functions

### Lambda Functions are one liner function defined by the Lambda Keyword  

## Syntax for Lambda Function

### lambda arguments:expression
### lambda --> keyword which simply let the Python Interpreter know that the current function is our Lambda Function.
### Arguments --> Input value for our Function.
### expression --> Logic (Function Body)

### Example --> lambda x:x**2

## They can have any number of Arguments but only one expression or only one logic.
## They are used for short operations or as Arguments to higher order Function.

## It can be treated as a Function without a Name and they dont even require a Return statement.

```python
# Function which calculates the square of the numbers using normal function

def square(num):
    return num**2
```

```python
square(5)
```

```python
# This similar problem can be solved using Lambda Function as well

lambda_variable = lambda x:x**2
```

```python
type(lambda_variable)
```

```python
lambda_variable(6)
```

```python
# Lambda Function to calculate the sum of two numbers

sum = lambda num1,num2:num1+num2
```

```python
sum(6,8)
```

```python
# product of multiple numbers

product = lambda a,b,c,d:a*b*c*d
print(product(1,2,3,5))
```
