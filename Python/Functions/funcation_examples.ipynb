```python
# 2. Define a function that would behave as Password Strength Checker

''' 
Strong password -->
if you password
    1. length is greater than 8 characters
    2. password must consist of atleast a lowercase (a,b,c,d,e..)
    3. password must consist of atleast a uppercase (A,B,C,D,E..)
    4. password must consist of atleast a digit (1,2,3,4,5)
    5. password must consist of atleast a special character (!@#$%^&)

Weak password -->
If a password doesn't match above condition then it will considered as weak password
''' 
```

### islower() --> checks that whether if the current character is a lowercase character or not.
### isupper() --> checks that whether if the current character is a uppercase character or not.
### isdigit() --> checks that whther if the current character is a digit or not.



```python
def password_strength(password):

    if len(password)<8:
        print(f'Entered {password} is a weak password')

    elif not any(char.islower() for char in password):
        print(f"Entered {password} is a week password")

    elif not any(char.isupper() for char in password):
        print(f"Entered {password} is a week password")

    elif not any(char.isdigit() for char in password):
        print(f"Entered {password} is a week password")

    elif not any(char in '!@#$%^&*()_' for char in password):
        print(f'Entered {password} is a week password')

    else:
        print(f'Password is strong')
```

```python
# 3. Check the string is Palindrome or not
# Palindrom --> A palindrome is a word, number, or string that reads the same forward and backward.
# eg --> madam → madam , racecar → racecar , 121 → 121 , level → level
# use reversing a string --> string[::-1]

def Palindrome(string):

    string_value = string.lower().replace(" ", "")

    if string_value == string_value[::-1]:
        print(f'{string} : is a Palindrome')
    
    else:
        print(f'{string} : is not a Palindrome')
```

```python
Palindrome('madam')
```

```python
Palindrome('ab a')    # space has been replaced with no space i.e aba = aba
```

```python
Palindrome('Aba')       # we have convert the string into lower that's why it is coming as Palnidrom i.e. aba = aba
```

```python
Palindrome('A man a plan a canal panama')
```

```python
# 4. Calculate the total cost of shopping cart

cart = [{'item1':'Apple' , 'Cost':5 , 'Quantity': 20},
        {'item2':'Banana' , 'Cost':4 , 'Quantity':15}, 
        {'item3':'Guava' , 'Cost':3 , 'Quantity':10}
]

# index in our cart list --> key name
# Tota --> Cost of 1 item * quantity

def cart_total(cart_items):
    Total_cost = 0
    for item in cart_items:
        Total_cost = Total_cost + item['Cost'] * item['Quantity']
        
    print(f"the total cost for the cart item is: {Total_cost}")
```

```python
cart_total(cart)
```
