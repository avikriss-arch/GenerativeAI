## File Handling 

## 1. Read the content from text file in the system

```python
# open() --> opening your file 
# First parameter --> File_name
# Second paramter --> Mode
# Mode --> It decides the way in which we are working with this file. It can 'r' or 'w'.
# Mode --> r --> Reading the content from the file
# Mode --> w --> Writing the content into the file
# Mode --> r+ --> Read the content first and then writing the content into the file.
# Mode --> w+ --> Writing the content first and then reading from the file.
```

```python
# Reading your File 
# read() --> Reads the content from the file 
```

```python
file_variable = open('example.txt', mode='r')          # step 1: open the file using open('file_name' , mode)
content_variable = file_variable.read()                # step 2: read the file using read() function
print(content_variable)                                # displaying the content 
file_variable.close()                                  # step 3: close the file using close() function
```

## 2. Write the content from the file

### write() --> it writes the content into the text file
### Mode --> w --> Content will be overwritten
### Mode --> a --> Content will be added from the end

```python
# write() --> helps us out in writing the content into the file. And it replaces the existing content and write a new content. To overcome this we use append mode (mode='a') just add new content from the next line of the content ends.
```

```python
file_variable = open('example.txt', mode='w')                           # step 1: open the file using open('file_name' , mode)
file_variable.write('we are now writing the content into our file')     # step 2: write into to file using write() function
file_variable.close()                                                   #step 3: close the file using close() function
```

```python
# Appending the content into the file
file_variable = open('example.txt' , 'a')
file_variable.write('\nI am learning GenAI pre-requisite Python')    # \n next line
file_variable.close()
```

## with --> It will be used for first of all opening the file as an Object and there will no requirement for closing the file.

```python
with open('example.txt' , 'r') as file:
    content_variable = (file.read())
    print(content_variable)
```

```python
with open('example.text' ,'w') as file:
    file.write('I am exploring Python to be expert in Generative AI')
```

```python
with open('example.text', 'a') as files:
    files.write('. I am exploring append mode as well in the same line.')
```

## Write the content into our file line by line using --> writelines() function

```python
list_lines = ['First line\n' , 'Second line\n' , 'Third line']
with open('example.txt' , 'w') as file1:
    file1.writelines(list_lines)
```

## Binary Format File

### Mode --> Write the content into the Binary File --> wb
### Mode --> Read the content from the Binary File --> rb

```python
data = b'\x00\x01\x02\x03\x04'
with open('example.txt' ,'wb') as binary_write:
    binary_write.write(data)
```

```python
with open('example.txt' ,'rb') as binary_read:
    binary_content = binary_read.read()
    print(binary_content)
```

```python
# Read the content from the source.txt file and then writing the content into an another file
with open('source_txt' , 'r') as source:
    content_read = source.read()


with open('destination.txt' , 'w') as destination:
    destination.write(content_read)
```

```python
# Reading the Total lines, Total words , Total character from a file using readlines() 

def count_lines(file):
    with open(file,'r') as files:
        lines = files.readlines()
        total_lines = len(lines)
        total_words = sum(len(l.split()) for l in lines)
        total_characters = sum(len(l) for l in lines)
    return total_lines , total_words , total_characters

file_path = 'total_data.txt'
Total_Lines, Total_Words , Total_Characters = count_lines(file_path)

print(f'Total lines are: {Total_Lines} \nTotal words are: {Total_Words} \nTotal characters are: {Total_Characters}')
```

### readlines() --> This readlines() function reads the line in a way where in it is like the lines are stored inside a lis['First_line' ,'Second_line' ,'Third_line']

```python
 # Write the content into a file and then after writing the content into that file, we are going to read the content from that file 
 # as well using seek() function
```

## w+ = First of all as W is there so it states that first writing operation will be done and then reading operation will be done.
## seek() --> It is going to help us out with reading the content after writing from the starting only 

```python
with open('total_data.txt' , 'w+') as write_read_file:
    write_read_file.write('I am using w+ example')             # existing data would be delete because are writing into the file. If we want to add the content with the existing content in the we have use append() function
    write_read_file.write('\nThis is second line of the file.')
    write_read_file.seek(0)
    content_variable_write_read = write_read_file.read()
    print(content_variable_write_read)
```
