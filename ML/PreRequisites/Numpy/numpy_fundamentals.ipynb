### -->Numpy Arrays facilitate advanceed Mathematical and other types of operation on large number of Data. Typically, such operations are executed more efficiently with less code than is possible using Python's built in sequences.

## Stacking and Spliting in Array
## --> Size of the Array must be same
## Stacking
## 1. Horizontal Stacking --> Row by Row joining --> hstack()
## 2. Vertical Stackign --> Column by Column joining --> vstack()
## Spliting
## 1. Horizontal Split --> hsplit(array, number of parts you want your array splited)
## 2. Vertical Split --> vsplit(array, number of parts)

```python
arr_stack1 = np.arange(1,13).reshape(3,4)
arr_stack1
```

```python
arr_stack2 = np.arange(13,25).reshape(3,4)
arr_stack2
```

```python
# Horizontal stacking

horizontal_array = np.hstack((arr_stack1,arr_stack2))       # we have to pass the value as a tuple so we used two parenthesis here otherwise it will throw error
horizontal_array
```

```python
horizontal_array = np.hstack((arr_stack1,arr_stack1))
horizontal_array
```

```python
horizontal_array = np.hstack((arr_stack2,arr_stack2))
horizontal_array
```

```python
horizontal_array = np.hstack((arr_stack1,arr_stack2,arr_stack1 ,arr_stack2))
horizontal_array
```

```python
# Vertical Stacking

vertical_array = np.vstack((arr_stack1,arr_stack2))
vertical_array
```

```python
horizontal_split = np.hsplit(vertical_array , 2)
horizontal_split
```

```python
horizontal_split = np.hsplit(vertical_array,4)
horizontal_split
```

```python
vertical_array
```

```python
vertical_split = np.vsplit(vertical_array, 2)
vertical_split
```

```python
vertical_split = np.vsplit(vertical_array,3)         # while spliting into 4 part it will throw error because it wont be divisible int 4 equal parts
vertical_split
```
