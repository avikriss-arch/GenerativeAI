# Assignment 2: User Menu (while loop + break/continue)
## Task 3

```python
orders = []             # empty list to store order_amount



while True:
    print("\n--- Menu ---")
    print("1 - Add order amount")
    print("2 - Show all orders and totals with discount")
    print("q - Quit")

    option = input("Enter your option 1/2/q : ")

    if option == '1':
        amount_input = input("Enter the amount: ")
        if amount_input.isdigit():
            orders.append(int(amount_input))
            print(f"Added order: {amount_input}")
        else:
            print("Invalid amount, please enter a number")
        continue

    elif option == '2':
        if len(orders) == 0:
            print("No order yet")
            continue

        print("\norder amount --> discount% --> final amount")
        total_revenue = 0

        for order_amount in orders:
            if order_amount>=2000:
                discount_amount = 15
            elif order_amount>=1500:
                discount_amount = 10
            elif order_amount>=1000:
                discount_amount = 7
            else:
                discount_amount = 0
            
            final_amount = order_amount -  (order_amount * discount_amount / 100)
            total_revenue += final_amount

            print(f"{order_amount}  -->  {discount_amount}%  -->  {final_amount}")


        print(f"\nTotal revenue after discount: {total_revenue}")
        continue

    elif option == 'q':
        print("Exit successfully")
        break

    else:
        print("Try again, Invalid option, please select from the available options 1/2/q only.")
        continue

```
