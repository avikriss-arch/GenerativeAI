```python
# 1.
price_dict = {
    "Laptop":70000,
    "Mobile":25000,
    "Keyboard":1000,
    "Mouse":800,
    "Headphones":3500,
    "Monitor":12000
}

type(price_dict)
```

```python
#2 
# Adding a new product with price
price_dict["Tablet"] = 30000
print(price_dict)
```

```python
# update the price of an existing product
price_dict["Mobile"] = 27000
print(price_dict)
```

```python
# Remove product by name
removed_product = price_dict.pop("Mouse", None)     #If Mouse exists remove it. if it doesn't just return None
print(removed_product)
```

```python
removed_product = price_dict.pop("Nokia" , None)
print(removed_product)
```

```python
#3. Average price of all products
print(price_dict)
```

```python
Total_sum_of_values = sum(price_dict.values())              # total sum of all values of price_dict dictionary
print(Total_sum_of_values)
```

```python
Total_length_of_values = len(price_dict)                    # total sum of all values of price_dict dictionary
print(Total_length_of_values)
```

```python
Average = Total_sum_of_values/Total_length_of_values        # Average of values of price_dict dictionary
print(Average)
```

```python
# Maximum price
max_price = max(price_dict.values())
print(max_price)
```

```python
# Minimum price
min_price = min(price_dict.values())
print(min_price)
```
