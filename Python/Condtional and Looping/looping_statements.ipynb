# Looping Statements

## Looping statement can be used for iterating over a series of numbers or through iterables or Data Structures or through a collection of data.

## range() 

### range() function can be used for working as our Iterable. 

### range() takes an input parameter and the first parameter that the range will be taking as the input will be the initial value that you want to print or you want to operate on and then you will be passing the next input after comma as the parameter as the final value and it can take third parameter as well and that will be the step size and that is the difference between the iterating values

```python
# Print the numbers from 1 to 50
# range(initial_value, final_value+1, step_size)

for item in range(1,51,1):      # gap size(step size) is 1
    print(item)
```

```python
for item in range(1,51,2):       # gap(step size) of 2 between the numbers
    print(item)
```

## In the range() there is never a necessity for mentioning the step size and similarly if you want to print the numbers or you want to operate on numbers again there is no necessity for passing the initial value as well if and only if your initial value is 0

```python
# numbers from 0 to 100
for item in  range(101):
    print(item)
```

```python
# To print the numbers from 0 to 100 with gap of 5
for item in range(0,101,5):
    print(item)
```

### By using range() we can represent the numbers in a decrementing manner as well or in a decreasing order as well 

```python
# Print the numbers from 50 to 1 with the Gap of 1 in a decrementing manner
for item in range(50,0,-1):        # -1 is used for gap of 1 in decrementing manner
    print(item)
```

```python
# Print the number from 50 to 0 with the gap of 5
for item in range(50,-1,-5):
    print(item)
```

## In Looping Statements , we can Iterate through Iterable like String as well 

```python
Name = "Avinash Singh"
for i in Name:
    print(i)
```

# Loop Control Statements

## Loop Control Statements are being used for terminating a loop in between when a condition is met or if you want to have some control over your iterating process or while iterating over the series of numbers

## break
### break is our first loop control statement which will help you out to terminate the loop when a certain condition is met

```python
# Print the numbers from 0 to 50 but once number=18 then terminate the loop or do not print either 18 or any number forward

for i in range(51):
    if i==18:
        break
    else:
        print(i)
```

## continue
### continue is our second loop control statement which helps us in iterating over a series of numbers but helps us with Skipping the value in between once we meet the condition.

```python
# Print the numbers from 0 to 20 but only print even numbers and for odd numbers skip those values

for i in range(21):
    if i%2==0:
        print(i)
    else:
        continue
```

```python
# print the numbers from 0 to 20 but once 17 is occuring skip it

for i in range(21):
    if i==17:
        continue
    else:
        print(i)
```

## pass
### pass statement is used for doing nothing

```python
# print the numbers from 0 to 20 but once 17 is occuring do nothing

for i in range(20):
    if i==17:
        pass
    else:
        print(i)
```

## Nested Loops
### It is going to have a Loop inside a Loop

### Nested loop can help us out with iterating over a series of numbers as with the proper combination of iterating numbers 

### f string --> f string will be helping us out with printing the numbers with the variable value present inside the colons so we can directly represent the variable value inside the colons message itself where variable value will be shown inside the curly braces {}

```python
for i in range(5):
    for j in range(4):
        print(f"i value is: {i} and j value is: {j}")
```

```python
# Print the sum of first N natural numbers where Natural numbers -->1,2,3...Nth

n = int(input("Enter the n natural number: "))

sum=0
for i in range(n+1):
    sum+=i
print(f"The sum of {n} natural number is :", sum)
```
