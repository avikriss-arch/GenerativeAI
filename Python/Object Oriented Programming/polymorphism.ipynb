# Polymorphism

### -->Pollymorphism means Many Forms in Greek
### -->In OOPs, Polymorphism allows objects of different classes as if they are the object of the same class and it enables reusability and makes program more flexible

## Types of Polymorphism
### 1.Method Overriding (Runtime Polymorphism)
### 2.Method Overloading (Compiletime Polymorphism)
### 3.Operator Overloading

## 1.Method Overriding
### --> A child class provides specific implementation of a method that has already been defined into the parent class.
### --> This allows dynamic selection at Runtime.

```python
class Grandparent:          # parent class

    def speak(self):
        print("Hello, My name is Avinash")

class Parent(Grandparent):      # child class
    
    def speak(self):
        print("Hello, My name is Avinash Jr.")

class Child(Grandparent):

    def speak(self):
        print("Hello, My name is Avinash junior junior")

```

```python
def attribute(objects):
     print(objects.speak())
    # return objects.speak()     #  --> If we return the value instead of printing it, 'none' will not come in output
```

```python
grandparent_obj = Grandparent()         # Grandparent object created
```

```python
print(attribute(grandparent_obj))
```

```python
parent_obj = Parent()
```

```python
print(attribute(parent_obj))
```

```python
child_obj = Child()
```

```python
print(attribute(child_obj))
```

```python
# example 2
class Shape:

    def area(self):
        return "This method returns area of the shape"
```

```python
class Rectangle(Shape):
    def __init__(self,width,height):
        self.width=width
        self.height=height

    def area(self):
        return self.width*self.height
```

```python
class Circle(Shape):
    def __init__(self,radius):
        self.radius=radius

    def area(self):
        return 3.14*self.radius*self.radius
```

```python
shape_obj = Shape()
print(shape_obj.area())
```

```python
rectangle_obj = Rectangle(4,4)
print(rectangle_obj.area())
```

```python
circle_obj = Circle(10)
print(circle_obj.area())
```

```python
# Example 3
class Agent:
    def respond(self):
        print('Generic agent response')

class SupportAgent(Agent):
    def respond(self):
        print('Resolving your support ticket')

class SalesAgent(Agent):
    def respond(self):
        print("Here's a product recommendation for you")



agents = [Agent() ,SupportAgent() ,SalesAgent()]            # object created for all three classes

for agent in agents:
    agent.respond()
```

```python
# Polymorphism via Duck Typing (Python-specific, very important) -->Python doesn't check if objects are "officially" related through inheritance
class Dog:
    def speak(self):
        print("Woof!")

class Cat:
    def speak(self):
        print("Meow!")

class Duck:
    def speak(self):
        print("Quack!")


animal_obj = [Dog(),Cat(),Duck()]       # Not related Dog Cat via Inheritance

for animal in animal_obj:
    animal.speak()                  # Works same as above, even though these classes are unrelated
```
