# Operators

## Operators helps in applying operations on the given operands.

```python
a=5
b=6
print(a+b)
```

## Types of operators:
1. Arithmetic operator

Addition(+)
Substraction(-)
Product or Multiplication (*)
Division (/)
Modulus (%)
Floor (//)
Exponent (a**b)

2. Comparison operators

Greater than (>)
Less than (<)
Greter than or equal to (>=)
Less than or equal to (<=)
Equality check (==)
Not equal to (!=)

3. Logical operator

and --> whether if both the conditions are true or not
or -->whether if any of the operation is true or not
not -->negation kind or to inverse the result of the given operations.

```python
# simple calculator

a=int(input("Enter first number: "))
b=int(input("Enter second number: "))
operation= input("Enter any of the operations: +, -, *, /")

if operation=="+":
    print("Addition of a and b is: ", a+b)
elif operation=="-":
    print("Subtraction of a and b is: ", a-b)
elif operation=="*":
    print("Multiplication of a and b is: ", a*b)
elif operation=="/":
    print("Division of a and b is: ", a/b)
```
